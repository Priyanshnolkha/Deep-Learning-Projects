```python
from ultralytics import YOLO
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

DATA_YAML = PROJECT_ROOT / "data" / "project_data.yaml"

MODEL_PATH = PROJECT_ROOT / "yolo26n.pt"

RESULTS_DIR = PROJECT_ROOT / "results"

print("Project root:", PROJECT_ROOT)
print("Dataset:", DATA_YAML)
print("Model:", MODEL_PATH)
```

```python
model = YOLO(str(MODEL_PATH))

print("YOLO26n loaded successfully.")
```

```python
results = model.train(
    data=str(DATA_YAML),

    epochs=30,
    imgsz=640,
    batch=4,

    device="mps",
    workers=0,

    # Augmentation
    hsv_h=0.015,
    hsv_s=0.4,
    hsv_v=0.2,

    degrees=5.0,
    translate=0.1,
    scale=0.5,
    shear=2.0,

    fliplr=0.5,
    flipud=0.0,

    mosaic=1.0,
    mixup=0.1,

    project=str(RESULTS_DIR),
    name="augmented_yolo26n",
    exist_ok=True
)
```
