```python
import pandas as pd

baseline = {
    "Precision": 0.8669765690907907,
    "Recall": 0.7337984234795751,
    "mAP50": 0.7964639849857705,
    "mAP50-95": 0.6286039015908174
}

# Replace these four values with your Model B results
augmented = {
    "Precision": 0.9000,
    "Recall": 0.8132,
    "mAP50": 0.8865,
    "mAP50-95": 0.6620
}

comparison = pd.DataFrame(
    [baseline, augmented],
    index=["Baseline YOLO26n", "Augmented YOLO26n"]
)

comparison
```

```python
improvement = pd.DataFrame({
    "Metric": [
        "Precision",
        "Recall",
        "mAP50",
        "mAP50-95"
    ],
    "Baseline": [
        baseline["Precision"],
        baseline["Recall"],
        baseline["mAP50"],
        baseline["mAP50-95"]
    ],
    "Augmented": [
        augmented["Precision"],
        augmented["Recall"],
        augmented["mAP50"],
        augmented["mAP50-95"]
    ]
})

improvement["Change (%)"] = (
    (improvement["Augmented"] - improvement["Baseline"])
    * 100
)

improvement
```

```python
import matplotlib.pyplot as plt

metrics_names = ["Precision", "Recall", "mAP50", "mAP50-95"]

baseline_values = [
    baseline["Precision"],
    baseline["Recall"],
    baseline["mAP50"],
    baseline["mAP50-95"]
]

augmented_values = [
    augmented["Precision"],
    augmented["Recall"],
    augmented["mAP50"],
    augmented["mAP50-95"]
]

x = range(len(metrics_names))

plt.figure(figsize=(10, 5))

plt.bar(
    [i - 0.2 for i in x],
    baseline_values,
    width=0.4,
    label="Baseline YOLO26n"
)

plt.bar(
    [i + 0.2 for i in x],
    augmented_values,
    width=0.4,
    label="Augmented YOLO26n"
)

plt.xticks(x, metrics_names)
plt.ylabel("Score")
plt.title("Baseline vs Augmented YOLO26n")
plt.legend()

plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```
